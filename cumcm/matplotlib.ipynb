```python
# draw plot
import matplotlib.pyplot as plt
from matplotlib import rcParams
# rcParams['font.family'] = 
# set size
plt.figure(figsize = (10, 5))
# plt.style.use('dark_background')

# data to be draw
month = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [100, 150, 80, 130]

# draw the plot
plt.plot(month, sales, label='Product A', color = 'green', linewidth = 2, linestyle = '--', marker = 'o')

# title
plt.title('The Trend of Sales in 2025', color = 'green')

# label of the axis
plt.xlabel('month', fontsize = 10)
plt.ylabel('sales', fontsize = 10)

# legend
plt.legend(loc='upper left')

# add grid
plt.grid(axis = 'y', alpha = 0.5, color = 'green', linestyle = '-')

# font size
plt.xticks(rotation = 0, fontsize = 14)

# set the range of Y axis
plt.ylim(0, 200)

# every data points show the values
for x, y in zip(month, sales):
    print(x, y)
    plt.text(x, y + 1, str(y), ha = 'center', va = 'bottom')
# show graph
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize = (10, 5))

subjects = ['Chinese', 'Math', 'English', 'Science']
scores = [85, 92, 78, 88]

plt.bar(subjects, scores, label = 'Lin')

plt.title('The grade distribution of the 2025', color = 'green', fontsize = 20)

plt.xlabel('subjects', fontsize = 10)
plt.ylabel('score', fontsize = 10)

plt.legend(loc = 'upper left')

plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize = (10, 5))

subjects = ['Chinese', 'Math', 'English', 'Science']
scores = [85, 92, 78, 88]

plt.barh(subjects, scores, label = 'Lin')

plt.title('The grade distribution of the 2025', color = 'green', fontsize = 20)

plt.xlabel('subjects', fontsize = 10)
plt.ylabel('score', fontsize = 10)

plt.legend(loc = 'upper left')

plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

things = ['study', 'entertainment', 'sports', 'sleep', 'other']
times = [6, 4, 1, 8, 5]
colors = ['']

plt.pie(times, labels = things, 
        autopct = '%.2f%%', 
        startangle=90,
        # colors = 'colors',
        wedgeprops={'width': 0.6},
        pctdistance=0.6,
        # explode = 'explode'
        )

plt.title('The distribution of a day time', color = 'green', fontsize = 20)    
plt.text(0, 0, 'Total: \n100%', ha = 'center', va = 'bottom', fontsize = 10)

plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt
import random
plt.figure(figsize=(10, 8))

plt.title('The relationship of the X variable and Y variable', color = 'green', fontsize = 20)
scores = [50, 55, 60, 65, 70, 75, 80]
hours = [1, 2, 3, 4, 5, 6, 7]

x = []
y = []
for i in range(1000):
    tmp = random.uniform(0, 10)
    x.append(tmp)
    y.append(2 * tmp + random.gauss(0, 2))

plt.scatter(x, y, color = 'green')

plt.ylim(0, 30)
plt.plot([0, 10], [0, 20])
plt.show()
```

```python
import matplotlib.pyplot as plt

data = {
    'Chinese': [82, 85, 88, 70, 90, 76, 84, 83, 95], 
    'Math': [75, 80, 79, 93, 88, 82, 87, 89, 92], 
    'English': [70, 72, 68, 65, 78, 80, 85, 90, 95]
}

plt.figure(figsize = (8, 6))
plt.boxplot(data.values(), tick_labels=data.keys())

plt.title('The distribution of the subjects(boxing)')
plt.ylabel('Scores')
plt.grid(True, axis = 'y', linestyle = '--', alpha = 0.5)
plt.show()
```

```python
# many plotting
import matplotlib.pyplot as plt
month = ['1', '2', '3', '4']
sales = [100, 150, 80, 130]
f1 = plt.subplot(2, 2, 1)
f1.plot(month, sales)
f2 = plt.subplot(2, 2, 2)
f2.bar(month, sales)
f3 = plt.subplot(2, 2, 3, alpha = 0.5)
f2.scatter(month, sales)
f4 = plt.subplot(2, 2, 4)
f4.barh(month, sales)
```
