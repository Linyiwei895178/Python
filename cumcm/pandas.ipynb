### DataFrame

```python
# the construction of the dataframe
import pandas as pd
import numpy as np
# constrct series through series
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([6, 7, 8, 9, 10])
df = pd.DataFrame({'The fist column:': s1, 'the second column:': s2})
type(df)
# construct df through dictionary
df = pd.DataFrame(
    {
        "name": ['tom', 'jack', 'alice', 'bob', 'allen'],
        "age": [15, 17, 20, 26, 30], 
        "score": [60.5, 80, 30.6, 70, 83.5]
    }, index = [1, 2, 3, 4, 5], columns=['name', 'score', 'age']
)
print(df)
# the property of the dataframe
print('row index: ')
print(df.index)
print('column label: ')
print(df.columns)
print(df.ndim)
print(df.dtypes)
# the transpose 
print(df.T)
print(df.T.index)
print('\n')
print(df.loc[4])
print(df.iloc[3])
```

```python
# some column 
print(df.loc[: ,'name'])
print(df.at[3, 'score'])
print(df.iat[2, 1])
print(df['name'])
print(type(df['name']))
print(df.head(2))
df.score > 70
df.sample(3)
```

```python
# common methods
print(df.head())
print(df.tail(1))
print(df.isin(['jack', 20])) # view the target element is or not included in the parameter set
print(df.isna())
print(df.score.max())
print(df.age.min())
print(df.score.mean())
print(df.score.median())
print(df.score.mode())
```

```python
print(df.score.std()) # standard var
print(df.score.var())
print(df.score.quantile(0.25))
print(df.describe())
```

```python
print(df.count()) # every column NOT NaN
print(df.value_counts())
print(df.duplicated(subset = ['score']))
df.sample(2)
df.cumsum()
df.cummax()
print(df.sort_index(ascending=False))
print(df.sort_values(by = 'name'))
```

```python
# case 1
data = {
    'Name': ['zhangsan', 'lisi', 'wangwu', 'zhaoliu', 'qianqi'],
    'Math': [85, 92, 78, 88, 95],
    'English': [90, 88, 85, 92, 80], 
    'Physics': [75, 80, 88, 85, 90]
}
scores = pd.DataFrame(data)
scores
# total and average
scores['Sum'] = scores[['Math', 'English', 'Physics']].sum(axis = 1)
scores['Average'] = scores['Sum'] / 3
scores[ (scores['Math'] > 90 ) | (scores['English'] > 85 )]
r1 = scores.sort_values('Sum', ascending = False).head(3)
r2 = scores.nlargest(3, columns = ['Sum'])
print(r1)
print(r2)
```
