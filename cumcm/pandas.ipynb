### 1.the construction of the series

```python
import pandas as pd
s = pd.Series([10, 2, 3, 4, 5])
print(s)
# defined index
s = pd.Series([10, 2, 3, 4, 5], index = ['A', 'B', 'C', 'D', 'E'])
print(s)
s = pd.Series([10, 2, 3, 4, 5], index = ['A', 'B', 'C', 'D', 'E'])
# construct by dictionary
s = pd.Series({"a": 1, "b": 2, "c": 3, "d": 4, "e": 5})
print(s)
#print(s)
# s = pd.Series(s, index = ["A", "C"])
# print(s)
```

```python
# Index:
s = pd.Series({"a": 1, "b": 2, "c": 3, "d": 4, "e": 5})
print(s.index)
print(s.values)
print(s.shape, s.ndim, s.size)
print(s.dtype, s.name)
# print(s.loc['c'])
print(s.iloc[1])
# print(s.at['a'])
# print(s.iat['a'])
```
