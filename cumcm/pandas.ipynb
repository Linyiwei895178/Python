### 1.the construction of the series

```python
import pandas as pd
s = pd.Series([10, 2, 3, 4, 5])
print(s)
# defined index
s = pd.Series([10, 2, 3, 4, 5], index = ['A', 'B', 'C', 'D', 'E'])
print(s)
s = pd.Series([10, 2, 3, 4, 5], index = ['A', 'B', 'C', 'D', 'E'])
# construct by dictionary
s = pd.Series({"a": 1, "b": 2, "c": 3, "d": 4, "e": 5})
print(s)
#print(s)
s = pd.Series(s, index = ["A", "C"])
print(s)
```
