```python
import pandas as pd
import numpy as np
np.random.seed(42)
values = np.random.randint(50, 101, 10)
indexes = []
for i in range(1, 11):
    indexes.append('student' + str(i))
scores = pd.Series(values, indexes)
print(scores)
print('average: ', scores.mean())
print('max: ', scores.max())
print('min: ', scores.min())
mean = scores.mean()
```

```python
temperatures = pd.Series([28, 31, 29, 32, 30, 27, 33], index = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'])
n = temperatures[temperatures > 30].count()
print('the days that above the 30 degrees:', n)
print('average degree: ', temperatures.mean())
t2 = temperatures.sort_values(ascending=False)
print('from high to low: ')
print(t2)
# the change is great
t3 = temperatures.diff().abs()
# diff is a common function
t3.sort_values(ascending=False).index
print('the 2:', t3.sort_values(ascending=False).keys()[:2].tolist())
```
