```python
import numpy as np

# -------------------------
# 1. 指标权重
# -------------------------

A = np.array([
    0.30,
    0.20,
    0.35,
    0.15
])

# -------------------------
# 2. 模糊关系矩阵
# -------------------------

R = np.array([
    [0.05, 0.10, 0.25, 0.45, 0.15],
    [0.10, 0.20, 0.40, 0.25, 0.05],
    [0.05, 0.10, 0.20, 0.40, 0.25],
    [0.10, 0.20, 0.35, 0.25, 0.10]
])

# -------------------------
# 3. 模糊综合评价
# -------------------------

B = A @ R

print("综合评价向量：")
print(B)

# -------------------------
# 4. 最大隶属度原则
# -------------------------

levels = [
    "差",
    "较差",
    "一般",
    "较好",
    "好"
]

index = np.argmax(B)

print("最终评价等级：", levels[index])

# -------------------------
# 5. 量化评分
# -------------------------

scores = np.array([
    20,
    40,
    60,
    80,
    100
])

final_score = B @ scores

print("综合得分：", final_score)
```

```python
import numpy as np
import pandas as pd

def fuzzy_comprehensive_evaluation(weights, relation_matrix, score_vector=None):
    """
    模糊综合评价函数 (基于加权平均型算子 M(·, +))
    :param weights: 权重向量 (一维数组或列表)
    :param relation_matrix: 隶属度矩阵 (m x n 维数组)
    :param score_vector: 各评语等级量化分值向量 (可选)
    :return: 归一化后的综合隶属度向量, 综合得分
    """
    W = np.array(weights)
    R = np.array(relation_matrix)
    
    # 1. 权重归一化校验
    if not np.isclose(np.sum(W), 1.0):
        W = W / np.sum(W)
        
    # 2. 模糊合成计算 B = W * R
    B = np.dot(W, R)
    
    # 3. 结果归一化
    B_normalized = B / np.sum(B)
    
    # 4. 计算综合得分 (若提供评语分值)
    final_score = None
    if score_vector is not None:
        S = np.array(score_vector)
        final_score = float(np.dot(B_normalized, S))
        
    return B_normalized, final_score

# --- 案例数据执行 ---
if __name__ == "__main__":
    # 指标与评语定义
    factors = ['教学态度', '教学内容', '教学方法', '教学效果']
    comments = ['优秀', '良好', '中等', '较差']
    
    # 1. 权重向量
    W = [0.20, 0.30, 0.25, 0.25]
    
    # 2. 单因素评价矩阵 (4x4)
    R = [
        [0.40, 0.40, 0.20, 0.00],
        [0.50, 0.30, 0.20, 0.00],
        [0.20, 0.50, 0.20, 0.10],
        [0.30, 0.40, 0.20, 0.10]
    ]
    
    # 3. 评语等级设定分值
    scores = [95, 80, 65, 50]
    
    # 执行评价
    B, final_score = fuzzy_comprehensive_evaluation(W, R, scores)
    
    # 格式化输出
    result_df = pd.DataFrame([B], columns=comments, index=['综合隶属度'])
    print("=== 模糊综合评价向量 ===")
    print(result_df.round(4))
    
    max_idx = np.argmax(B)
    print(f"\n最大隶属度评价结果: {comments[max_idx]} (隶属度: {B[max_idx]:.4f})")
    print(f"最终加权量化得分: {final_score:.3f} 分")
```
