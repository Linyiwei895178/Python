```python
from scipy.optimize import linprog

# 最大化：
# Z = 30x1 + 20x2
#
# scipy.linprog 默认求最小值
# 所以目标函数取负号

c = [-30, -20]

# 约束：
# 2x1 + x2 <= 100
# x1 + 2x2 <= 80

A = [
    [2, 1],
    [1, 2]
]

b = [100, 80]

# x1 >= 0
# x2 >= 0

bounds = [
    (0, None),
    (0, None)
]

result = linprog(
    c,
    A_ub=A,
    b_ub=b,
    bounds=bounds,
    method="highs"
)

print("是否成功：", result.success)
print("最优解：", result.x)
print("最大利润：", -result.fun)
```

```python
import pulp

# 创建最大化问题
model = pulp.LpProblem(
    "Project_Selection",
    pulp.LpMaximize
)

# 决策变量
x = [
    pulp.LpVariable(
        f"x{i}",
        cat="Binary"
    )
    for i in range(5)
]

cost = [10, 15, 12, 8, 20]
profit = [20, 30, 25, 16, 40]

# 目标函数
model += pulp.lpSum(
    profit[i] * x[i]
    for i in range(5)
)

# 预算约束
model += pulp.lpSum(
    cost[i] * x[i]
    for i in range(5)
) <= 35

# 求解
model.solve()

print("状态：", pulp.LpStatus[model.status])

for i in range(5):
    print(
        f"项目 {i + 1}:",
        pulp.value(x[i])
    )

print("最大收益：", pulp.value(model.objective))
```

```python
from scipy.optimize import minimize

def objective(x):
    return -(100 * x[0] - x[0] ** 2)

# 约束 0 <= x <= 100
bounds = [
    (0, 100)
]

result = minimize(
    objective,
    x0=[20],
    bounds=bounds,
    method="SLSQP"
)

print("是否成功：", result.success)
print("最优生产量：", result.x[0])
print("最大利润：", -result.fun)
```

```python
import numpy as np
from scipy.optimize import minimize

def objective(x):
    x1, x2 = x

    return -(
        40 * x1
        + 30 * x2
        - x1 ** 2
        - x2 ** 2
    )

# x1 + x2 <= 40
constraints = [
    {
        "type": "ineq",
        "fun": lambda x: 40 - x[0] - x[1]
    }
]

bounds = [
    (0, None),
    (0, None)
]

result = minimize(
    objective,
    x0=np.array([10, 10]),
    bounds=bounds,
    constraints=constraints,
    method="SLSQP"
)

print("是否成功：", result.success)

x1, x2 = result.x

print("x1 =", x1)
print("x2 =", x2)
print("最大利润 =", -result.fun)
```

```python
import numpy as np
from scipy.optimize import linprog

# 1. 目标函数系数 (求最大化需要取负)
c = np.array([-2, -3]) 

# 2. 不等式约束的系数矩阵 (A_ub * x <= b_ub)
A_ub = np.array([
    [1, 2],  # 甲材料约束
    [4, 0],  # A的工时约束
    [0, 4]   # B的工时约束
])
b_ub = np.array([8, 16, 12])

# 3. 变量边界 (默认就是 >= 0，但显式写出更好)
x0_bounds = (0, None)
x1_bounds = (0, None)

# 4. 求解
result = linprog(c, A_ub=A_ub, b_ub=b_ub, bounds=[x0_bounds, x1_bounds], method='highs')

print(f"最优方案: A产量={result.x[0]:.2f}吨, B产量={result.x[1]:.2f}吨")
print(f"最大利润: {-result.fun:.2f}万元")
```

```python
import numpy as np
from scipy.optimize import milp, LinearConstraint, Bounds

# 目标函数系数 (取负求最大化)
c = np.array([-10, -15, -22, -17])

# 约束条件 5x1 + 10x2 + 12x3 + 8x4 <= 20
A = np.array([[5, 10, 12, 8]])
constraints = LinearConstraint(A, ub=20) # ub为上限上限(upper bound)

# 定义变量的类型：1 代表整数变量
integrality = np.ones_like(c)

# 限制变量在 0 和 1 之间
bounds = Bounds(0, 1)

# 求解
result = milp(c=c, constraints=constraints, integrality=integrality, bounds=bounds)

print(f"投资选择 (1为投, 0为不投): {result.x}")
print(f"最大回报: {-result.fun:.2f}万元")
```

```python
import numpy as np
from scipy.optimize import minimize

# 1. 定义目标函数 (传入的参数 vars 是一个包含 x 和 y 的数组)
def objective(vars):
    x, y = vars[0], vars[1]
    dist_A = np.sqrt((x - 0)**2 + (y - 0)**2)
    dist_B = np.sqrt((x - 0)**2 + (y - 4)**2)
    dist_C = np.sqrt((x - 5)**2 + (y - 0)**2)
    return dist_A + dist_B + dist_C

# 2. 初始猜测值 (猜测选在三个点的重心附近)
x0 = [1.5, 1.5]

# 3. 求解 (默认使用 BFGS 算法)
result = minimize(objective, x0)

print(f"救援中心最佳坐标: x={result.x[0]:.2f}, y={result.x[1]:.2f}")
print(f"最小距离之和: {result.fun:.2f}")
```

```python
import pulp

# ================= 1. 准备数据 =================
warehouses = ['A', 'B', 'C']
customers = ['1', '2', '3', '4']

# 固定成本 (F_i) 和 容量 (M_i)
fixed_costs = {'A': 1000, 'B': 1200, 'C': 1500}
capacities = {'A': 500, 'B': 600, 'C': 800}

# 需求量 (D_j)
demands = {'1': 200, '2': 300, '3': 150, '4': 250}

# 运输成本矩阵 (C_ij)
transport_costs = {
    'A': {'1': 2, '2': 4, '3': 5, '4': 3},
    'B': {'1': 3, '2': 1, '3': 4, '4': 2},
    'C': {'1': 4, '2': 3, '3': 2, '4': 1}
}

# ================= 2. 建立模型 =================
# 创建一个求最小值的模型对象
model = pulp.LpProblem("Facility_Location_Problem", pulp.LpMinimize)

# ================= 3. 定义决策变量 =================
# y_i: 0-1 变量，是否建仓 (LpBinary代表整数0或1)
y = pulp.LpVariable.dicts("Build", warehouses, cat=pulp.LpBinary)

# x_ij: 连续变量，运输量 (lowBound=0代表不能是负数)
x = pulp.LpVariable.dicts("Transport", 
                          [(i, j) for i in warehouses for j in customers], 
                          lowBound=0, cat=pulp.LpContinuous)

# ================= 4. 添加目标函数 =================
# 目标 = 所有的固定成本 + 所有的运输成本
model += (
    pulp.lpSum([fixed_costs[i] * y[i] for i in warehouses]) + 
    pulp.lpSum([transport_costs[i][j] * x[(i, j)] for i in warehouses for j in customers])
)

# ================= 5. 添加约束条件 =================
# 约束一：满足每个片区的需求量
for j in customers:
    model += pulp.lpSum([x[(i, j)] for i in warehouses]) == demands[j], f"Demand_{j}"

# 约束二：逻辑绑定与容量限制（核心逻辑）
for i in warehouses:
    model += pulp.lpSum([x[(i, j)] for j in customers]) <= capacities[i] * y[i], f"Capacity_and_Logic_{i}"

# ================= 6. 求解与输出 =================
model.solve()

print(f"求解状态: {pulp.LpStatus[model.status]}\n")
print(f"最小总成本: {pulp.value(model.objective)} 万元\n")

print("--- 建仓决策 ---")
for i in warehouses:
    if y[i].varValue > 0.5: # 浮点数精度问题，通常用 > 0.5 判断是否为 1
        print(f"在 {i} 地建立仓库")

print("\n--- 运输方案 ---")
for i in warehouses:
    for j in customers:
        if x[(i, j)].varValue > 0:
            print(f"从仓库 {i} 运送 {x[(i, j)].varValue} 吨到片区 {j}")
```
